```python
import os
import pandas as pd
import ast
from collections import defaultdict
```

```python
# Set the directory where your CSV files are located
folder_path = 'data/user_review_subsets'

# Get a list of all CSV files in the folder
csv_files = [file for file in os.listdir(folder_path) if file.endswith('.csv')]

# Initialize an empty list to store individual dataframes
dfs = []

# Loop through each CSV file, read it into a dataframe, and append it to the list
for file in csv_files:
    file_path = os.path.join(folder_path, file)
    df = pd.read_csv(file_path)
    dfs.append(df)

# Concatenate all dataframes into one
user_review_df = pd.concat(dfs, ignore_index=True)
user_review_df['all_reviews'] = user_review_df['all_reviews'].str.replace('nan', '"nan"')
user_review_df['all_reviews'] = user_review_df['all_reviews'].apply(ast.literal_eval)
```

```python
user_review_df
```

```python
usernames = []
for i, row in user_review_df.iterrows():
    for review in row.all_reviews:
        usernames.append((review["username"]))
```

```python
usernames
```

```python
from collections import defaultdict

# Initialize a defaultdict to count occurrences of each string
string_counts = defaultdict(int)

# Count occurrences of each string
for string in usernames:
    string_counts[string] += 1

# Sort the dictionary based on counts in descending order
sorted_counts = sorted(string_counts.items(), key=lambda x: x[1], reverse=True)

# Print the strings and their counts in order
for string, count in sorted_counts:
    print(f"{string}: {count}")
```

### restructure user review dataframe

```python
user_review_df
```

```python
product_hrefs = []
user_hrefs = []
usernames = []
ratings = []
dates = []
for i, row in user_review_df.iterrows():
    for review in row.all_reviews:
        product_hrefs.append(row.href)
        user_hrefs.append(review["href"])
        usernames.append(review["username"])
        ratings.append(review["rating"])
        dates.append(review["date"])

user_review_df = pd.DataFrame({
    "product_href" : product_hrefs,
    "user_href" : user_hrefs,
    "username" : usernames,
    "rating" : ratings,
    "date" : dates,
})
```

```python
# drop users where username is blank
print("length before removal of missing usernames:", len(user_review_df))
user_review_df = user_review_df[user_review_df['username'] != ''].reset_index(drop=True)
print("length before after of missing usernames:", len(user_review_df))
```
